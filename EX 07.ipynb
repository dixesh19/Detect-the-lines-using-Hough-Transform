```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("C:\\Users\\admin\\Downloads\\lan_img1.jpg",0)
img_c=cv2.imread("C:\\Users\\admin\\Downloads\\lan_img1.jpg",1)
img_c=cv2.cvtColor(img_c,cv2.COLOR_BGR2RGB)
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_GRAY2RGB)
gray = cv2.GaussianBlur(gray,(3,3),0)
```

```python
print("DINESH R")
print("212224240037")
plt.figure(figsize=(13,13))
plt.subplot(1,2,1)
plt.imshow(img_c)
plt.title("Original Image")
plt.axis("off")
```

```python
print("DINESH R")
print("212224240037")
plt.subplot(1,2,2)
plt.imshow(gray)
plt.title("Gray Image")
plt.axis("off")
plt.show()
```

```python
print("DINESH R")
print("212224240037")
canny=cv2.Canny(gray,120,150)
plt.imshow(canny)
plt.title("Canny Edge Detector")
plt.axis("off")
plt.show()
```

```python
lines=cv2.HoughLinesP(canny,1,np.pi/180,threshold=80,minLineLength=50,maxLineGap=250)
```

```python
for line in lines:
    x1,y1,x2,y2=line[0]
    cv2.line(img_c,(x1,y1),(x2,y2),(255,0,0),3)
```

```python
print("DINESH R")
print("212224240037")
plt.imshow(img_c)
plt.title("Result Image")
plt.axis("off")
plt.show()
```
